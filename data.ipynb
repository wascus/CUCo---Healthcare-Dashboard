```python
import pandas as pd
import os
```

```python
path = "./Data"
```

```python
mental_health_and_technology = path+"/mental-health-and-technology/"
mental_wellness_tracker = path+"/mental-wellness-tracker/"
psychological_assessment = path+"/psychological-assessment-dataset/"
```

```python
mental_health_and_technology_csv_files = [f for f in os.listdir(mental_health_and_technology) if f.endswith(".csv")]
mental_wellness_tracker_csv_files = [f for f in os.listdir(mental_wellness_tracker) if f.endswith(".csv")]
psychological_assessment_csv_files = [f for f in os.listdir(psychological_assessment) if f.endswith(".csv")]
```

```python
mental_health_and_technology_datasets = {file: pd.read_csv(os.path.join(mental_health_and_technology, file)) for file in mental_health_and_technology_csv_files}
mental_wellness_tracker_datasets = {file: pd.read_csv(os.path.join(mental_wellness_tracker, file)) for file in mental_wellness_tracker_csv_files}
psychological_assessment_datasets = {file: pd.read_csv(os.path.join(psychological_assessment, file)) for file in psychological_assessment_csv_files}
```

```python
for name, df in mental_health_and_technology_datasets.items():
    print(f"--- {name} ---")
    print(df.head(), "\n")
```

```python
for name, df in mental_wellness_tracker_datasets.items():
    print(f"--- {name} ---")
    print(df.head(), "\n")
```

```python
for name, df in psychological_assessment_datasets.items():
    print(f"--- {name} ---")
    print(df.head(), "\n")
```
